```

import json

# Load the annotated data
with open('/mnt/data/annotated_skill_roles_spacy.json', 'r') as f:
    skill_roles_data = json.load(f)

with open('/mnt/data/annotated_course_details_spacy.json', 'r') as f:
    course_details_data = json.load(f)

# Display a sample from each dataset to understand the structure
sample_skill_roles = skill_roles_data[:2]  # Taking first two samples
sample_course_details = course_details_data[:2]  # Taking first two samples

(sample_skill_roles, sample_course_details)
```

### The data has been successfully loaded and consists of text segments with associated entities.

```

# Extract entities from the skill roles data
def extract_entities(data):
    extracted_entities = []
    for entry in data:
        text = entry[0]
        entities = entry[1]['entities']
        for entity in entities:
            start, end, label = entity
            extracted_entities.append((text[start:end], label))
    return extracted_entities

skill_roles_entities = extract_entities(skill_roles_data)
course_details_entities = extract_entities(course_details_data)

skill_roles_entities[:5], course_details_entities[:5]  # Display first 5 entities from each dataset
```
