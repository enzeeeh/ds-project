```python
from bs4 import BeautifulSoup
import requests
import csv
import pandas as pd
import time
```

Parse the HTLM Content

```python
url = 'https://www.masterstudies.com/masters-degree/cyber-security'
url2 = 'https://www.mygermanuniversity.com/studyfinder?p=1&pp=25&sort=featured&dir=ASC&sbj=68&skw=cyber'
url3 = 'https://books.toscrape.com/'

response = requests.get(url3)
soup = BeautifulSoup(response.text, 'html.parser')

if response.status_code == 200:
    print("request succesfull")
    print(soup.prettify())
else:
    print('failed')
```

```python
books = soup.find_all('h3')[0]
print(books)

# book_url = books.find('a')['href']
# book_respose = requests.get(url3 + book_url)
# book_soup = BeautifulSoup(book_respose.content, 'html.parser')
# print(book_soup.prettify())
```

```python
books = soup.find_all('h3')

start_time = time.time()
book_extracted = 0

for book in books:
    book_url = book.find('a')['href']
    book_respose = requests.get(url3 + book_url)
    book_soup = BeautifulSoup(book_respose.content, 'html.parser')
    
    title = book_soup.find('h1').text
    category = book_soup.find('ul', class_='breadcrumb').find_all('a')[2].text.strip()
    rating = book_soup.find('p', class_='star-rating')['class'][1]
    price = book_soup.find('p', class_='price_color').text.strip()
    availability = book_soup.find('p', class_='availability').text.strip()
    
    book_extracted += 1
    
    end_time = time.time()
    total_time = (end_time-start_time)/60.0
    
    print(f'Title: {title}')
    print(f'Category: {category}')
    print(f'Rating: {rating}')
    print(f'Price: {price}')
    print(f'Availability: {availability}')
    print('===================')
```

Extract detail from 50 pages

```python
books_data = []
for page_num in range(1,2):
    url4 = f'https://books.toscrape.com/catalogue/page-{page_num}.html'
    response = requests.get(url4)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    books = soup.find_all('h3')

    start_time = time.time()
    book_extracted = 0

    for book in books:
        book_url = book.find('a')['href']
        book_respose = requests.get('https://books.toscrape.com/catalogue/' + book_url)
        book_soup = BeautifulSoup(book_respose.content, 'html.parser')
        
        title = book_soup.find('h1').text
        category = book_soup.find('ul', class_='breadcrumb').find_all('a')[2].text.strip()
        rating = book_soup.find('p', class_='star-rating')['class'][1]
        price = book_soup.find('p', class_='price_color').text.strip()
        availability = book_soup.find('p', class_='availability').text.strip()
        
        book_extracted += 1
        
        end_time = time.time()
        total_time = (end_time-start_time)/60.0
        
        books_data.append([title,category,rating,price,availability])
        print(books_data)
        print('*********')
        print(f'{page_num*len(books)} books exracted so far...')
        
```

Export to CSV

```python
df = pd.DataFrame(books_data, columns=["Title","Category","Rating","Price","Availability"])
print(df.head)
```

```python
df.to_csv("books_scrapped.csv", index=False)
```
