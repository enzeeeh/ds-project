```python
# List of IDs
idx = ["TKIR24","TCSR24","TEHS24","TFCS24","TCOS25","TEKR23"]

# Initialize an empty list to store dictionaries of data
data_list = []

for id in idx:
    
    if id.endswith("24"):
    # Construct the URL
        web_url = f"https://ju.se/en/study-at-ju/courses.html?courseCode={id}&semester=20242"
    else:
        web_url = f"https://ju.se/en/study-at-ju/courses.html?courseCode={id}&semester=20251"
        
    # Access the URL and fetch data
    driver.get(web_url)
    
    # Fetch course title
    wait = WebDriverWait(driver, 20)
    course_title_element = wait.until(EC.visibility_of_element_located((By.CSS_SELECTOR, "h1.nomargin.c8632")))
    course_title = course_title_element.text

    # The course description seems to follow the "Course Contents" section,
    # We can grab all text in <div id="details"> and then process it if needed
    course_details_element = wait.until(EC.visibility_of_element_located((By.ID, "details")))
    course_contents_script = """
    var contentNode = document.querySelector('h2.c8632');
    var text = '';
    var nextNode = contentNode.nextSibling;
    while(nextNode && nextNode.nodeName.toLowerCase() !== 'h2') {
    if (nextNode.nodeType === 3) { // Node.TEXT_NODE
        text += nextNode.textContent.trim() + '\\n';
    }
    nextNode = nextNode.nextSibling;
    }
    return text;
    """

    course_desc = driver.execute_script(course_contents_script)
    
    entry_data = {
        "Course Title": course_title,
        "Course Description": course_desc
    }
    # Append this entry to the list
    data_list.append(entry_data)
        
# Convert the list of dictionaries to a DataFrame
df13 = pd.DataFrame(data_list)
df13
with pd.ExcelWriter('dataset.xlsx', engine='openpyxl', mode='a') as writer:
    # Write the new DataFrame to a new sheet
    df13.to_excel(writer, sheet_name='course-13', index=False)
    print("Successfully write to the excel")
```

```python
import time

# Initialize WebDriver
web_url = 'https://masters.ds.unipi.gr/security/en/category/digital-systems-security/'
 
# Navigate to your page
driver.get(web_url)

# Wait for the lesson_archive_post elements to load
wait = WebDriverWait(driver, 20)
lesson_posts = wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME, "lesson_archive_post")))

# Initialize lists to hold course titles and details
course_titles = []
course_details = []

# Iterate over each lesson_archive_post element
for post in lesson_posts:
    # Extract the course title
    title_element = post.find_element(By.CLASS_NAME, "lesson_archive_title_main")
    course_titles.append(title_element.text)
    
   # Attempt to click the element to reveal details
    # Adjust the selector as needed based on your page's behavior
    try:
        toggle_element = post  # Assuming clicking the post reveals the details; adjust if it's a specific element
        toggle_element.click()  # Click to reveal details
        time.sleep(1)  # Adjust based on how quickly the details are displayed

        # Now, extract the course details paragraphs
        details_elements = post.find_elements(By.XPATH, ".//div[contains(@class,'lesson_details_p')]/p")
        details_text = ' '.join([detail.text for detail in details_elements])
    except Exception as e:
        print(f"Error revealing details for {title_element.text}: {e}")
        details_text = "Details not found"

    course_details.append(details_text)

# Combine course titles and details into a DataFrame for easier handling
df14 = pd.DataFrame({
    'Course Title': course_titles,
    'Course Description': course_details
})

# Display the DataFrame
df14
with pd.ExcelWriter('dataset.xlsx', engine='openpyxl', mode='a') as writer:
    # Write the new DataFrame to a new sheet
    df14.to_excel(writer, sheet_name='course-14', index=False)
    print("Successfully write to the excel")
```
